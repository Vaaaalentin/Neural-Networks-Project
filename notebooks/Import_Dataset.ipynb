```python
import shutil
from pathlib import Path

selected_sports = [
    "football",
    "basketball",
    "tennis",
    "golf",
    "swimming",
    "boxing",
    "cycling",
    "surfing",
    "baseball",
    "ice hockey",
    "badminton",
    "gymnastics",
    "skiing",
    "rowing",
    "rugby",
]

source_dir = Path("sports_100")
target_dir = Path("sports_15")

for split in ["train", "valid", "test"]:
  for sport in selected_sports:
    src = source_dir / split / sport
    dst = target_dir / split / sport
    if src.exists():
      shutil.copytree(src, dst, dirs_exist_ok=True)

print("Terminé ! Votre dataset de 15 sports est prêt dans le dossier 'sports_15'.")
```
